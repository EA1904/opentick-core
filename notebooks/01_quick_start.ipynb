# TradoVera — Tutoriel 1 : Prise en main rapide du SDK

Ce notebook vous montre comment importer le SDK TradoVera en Python, charger les prix historiques, interroger le catalogue et exécuter des requêtes SQL DuckDB sur le Data Lake.

```python
import os
import sys

# Ajouter le répertoire parent pour pouvoir importer tvdata_sdk
sys.path.append(os.path.abspath(".."))

import tvdata_sdk
```

### 1. Charger des données de prix OHLCV
Nous pouvons charger l'historique quotidien d'un symbole en appelant simplement `get_ohlcv()` :

```python
df = tvdata_sdk.get_ohlcv("AAPL", timeframe="D1")
print(f"AAPL loaded: {len(df)} rows")
df.head()
```

### 2. Récupérer les métadonnées
Accédez facilement aux détails commerciaux et sectoriels enregistrés dans `catalog.db` :

```python
meta_df = tvdata_sdk.get_metadata("AAPL")
meta_df
```

### 3. Exécuter des requêtes SQL DuckDB sur le lac Parquet
Le SDK permet d'utiliser toute la puissance du SQL analytique directement sur les fichiers Parquet sous-jacents :

```python
query = """
SELECT 
    symbol, 
    count_star() as num_days, 
    min(timestamp) as min_date, 
    max(timestamp) as max_date 
FROM ohlcv 
WHERE symbol IN ('AAPL', 'MSFT', 'TSLA')
GROUP BY symbol
"""
tvdata_sdk.sql_query(query)
```
